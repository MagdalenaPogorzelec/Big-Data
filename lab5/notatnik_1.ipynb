Najważniejsze narzędzia analityczne / Big Data / AI w chmurach (gł. Azure)

**Big Data & Analytics (Azure)**
- Azure Data Lake Storage Gen2 – skalowalna przestrzeń danych do przechowywania danych strukturalnych i niestrukturalnych.
- Azure Synapse Analytics – platforma do integracji hurtowni danych i Big Data (SQL + Spark).
- Azure Databricks – oparty na Apache Spark, idealny do machine learning i przetwarzania dużych zbiorów danych.
- Azure Data Factory – narzędzie do tworzenia pipeline’ów ETL/ELT.
- Azure Stream Analytics – analiza danych strumieniowych w czasie rzeczywistym.
- Power BI – wizualizacja danych.

**AI/ML Tools (Azure)**
- Azure Machine Learning (AML) – platforma do budowy, trenowania i wdrażania modeli ML.
- Cognitive Services (Vision, Text, Speech) – gotowe API AI do analizy obrazów, tekstu, mowy itd.
- Azure OpenAI Service – dostęp do GPT-4 i innych modeli foundation.

**Compute & Orchestration**
- Azure Kubernetes Service (AKS) – uruchamianie kontenerów w chmurze.
- Azure Functions – serverless, dobre do uruchamiania event-driven logic.
- Event Grid / IoT Hub / Event Hub – przesyłanie danych z IoT i eventów do dalszego przetwarzania.



Przykład PoC: Wykrycie anomalii z linii produkcyjnej (IoT) – Azure

**Cel PoC**
- Wykrywanie anomalii w czasie rzeczywistym na podstawie danych z sensorów IoT z linii produkcyjnej.

**Architektura**
- IoT Hub – odbiera dane z urządzeń IoT (czujniki temperatury, drgań, itp.).
- Azure Stream Analytics – przetwarza dane w czasie rzeczywistym, wstępne agregacje, wykrywanie prostych wzorców.
- Azure Functions – reaguje na alerty/anomalie i zapisuje je do bazy danych lub wysyła powiadomienie.
- Azure Data Lake – długoterminowe przechowywanie danych do analiz historycznych.
- Azure Machine Learning – trenowanie modelu do wykrywania anomalii (np. Isolation Forest, AutoEncoder).
- Power BI – dashboard do wizualizacji danych i alertów.

**Proces PoC**
- Symulacja danych z IoT (lub realne dane).
- Zbudowanie pipeline'u do streamingu i zapisu danych.
- Trenowanie modelu ML na danych historycznych.
- Wdrożenie modelu jako web service w Azure ML.
- Połączenie modelu z Azure Stream Analytics.
- Walidacja wyników i iteracja.

Big Data Architektura dla Rockstar Games – Analiza UX i zapobieganie błędom w grze

**Cel**
- Zbieranie danych o zachowaniu graczy, błędach w grze i metrykach UX, ich analiza i wykrywanie potencjalnych problemów technicznych i projektowych.

**Architektura (Azure)**
- Data Ingestion
- Event Hubs / Azure Front Door + Functions – zbieranie danych z aplikacji (logi, błędy, kliknięcia).
- Azure IoT Hub – jeżeli są dane z urządzeń lub gamepadów (np. latency).

**Data Lake**
- Azure Data Lake Storage Gen2 – warstwa surowych danych.

**Processing & Storage**
- Azure Databricks – przetwarzanie danych, trenowanie modeli ML.
- Azure Synapse – analizy biznesowe, agregacje, SQL.
- Azure Cosmos DB / SQL Database – szybki dostęp do danych online.

**ML/AI** \
**Azure ML – modele do:**
- klasyfikacji błędów,
- predykcji rage-quitów,
- segmentacji graczy wg zachowania,
- analizy sentymentu z czatów.

**Reporting & Visualization**
- Power BI – dashboardy dla zespołu dev i product.

**Monitoring & Alerts**
- Azure Monitor + Log Analytics – śledzenie metryk i wykrywanie anomalii w systemie.

**Compute**
- Databricks (cluster Spark) – heavy ML processing.
- Azure ML Compute / AKS – serving modeli.
- Functions / Logic Apps – automatyzacja zadań.
