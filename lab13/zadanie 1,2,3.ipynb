##  Zadanie 2 – Delta Lake vs Apache Iceberg

| Kryterium           | Delta Lake                         | Apache Iceberg                          |
|---------------------|-------------------------------------|------------------------------------------|
| Popularność         | Wysoka (Databricks)                | Rosnąca (open-source)                    |
| ACID                | Tak                                | Tak                                      |
| Obsługa silników    | Spark, Databricks                  | Spark, Flink, Trino, Presto              |
| Ewolucja schematu   | Prosta                             | Bardziej zaawansowana                    |
| Wydajność zapytań   | Wysoka                             | Bardzo wysoka                            |
| Time travel         | Tak                                | Tak, rozbudowany katalog                 |
| Open Source         | Częściowo                          | Tak                                      |

###  Rekomendacja:

- **Delta Lake** – Gdy używasz Databricks i potrzebujesz prostoty.
- **Iceberg** – Gdy potrzebujesz interoperacyjności i rozbudowanych katalogów.
