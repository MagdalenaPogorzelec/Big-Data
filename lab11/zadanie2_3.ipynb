# ZADANIE 2

```
from pyspark.sql import SparkSession
filePath = "dbfs:/FileStore/tables/Files/movies.csv"
df = (spark.read.format("csv")
              .option("header","true")
              .option("inferSchema","true")
              .load(filePath))
display(df.limit(5))
```

1. Partitioned Write

```
df.write \
    .mode("overwrite") \
    .partitionBy("year") \
    .format("parquet") \
    .saveAsTable("default.partitioned_table")
```

 2. Bucketed Write


```
spark.sql("CREATE DATABASE IF NOT EXISTS demo_db")
spark.catalog.setCurrentDatabase("demo_db")

df.write \
    .format("parquet") \
    .bucketBy(8, "year") \
    .sortBy("year") \
    .mode("overwrite") \
    .saveAsTable("demo_db.bucketed_table")

```

 3. Porównanie tabel


```
print("Partitioned table:")
display(spark.table("default.partitioned_table").groupBy("year").count().orderBy("year"))

print("Bucketed table:")
display(spark.table("demo_db.bucketed_table").groupBy("year").count().orderBy("year"))
```
